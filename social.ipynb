```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
import numpy as np
```

```python
data_df=pd.read_csv("Social_media_impact_on_life-selected-columns.csv")
data_df
```

# HISTOGRAM

```python
mean_value = data_df["Daily_Usage_Hours"].mean()
median_value = data_df["Daily_Usage_Hours"].median()
mode_value = data_df["Daily_Usage_Hours"].mode()[0]

data_df.plot.hist(y="Daily_Usage_Hours", figsize = (12, 6), color = "cyan",bins = 20,edgecolor = "black", title = "DAILY USAGE HOURS", density = True)
data_df["Daily_Usage_Hours"].plot.kde(color = "Red")

plt.axvline(mode_value, color = "yellow", label = f"Mode: {mode_value}")
plt.axvline(median_value, color = "green", linestyle = "dotted", label = f"Median: {median_value}")
plt.axvline(mean_value, color = "grey", linestyle = "dotted", label = f"Mean: {mean_value: .2f}")

plt.legend()
plt.show()
```

## __POSITIVELY SKEWED__

### _Explanation_
The mean is greater than both the median and the mode.
The kernel Density Estimator tail is stretching to the right. 

```python
data_df.tail(5)
plt.show()
```

# SCATTER DIAGRAM

```python
data_df.plot.scatter(x = "Daily_Usage_Hours", y = "Academic_Performance_GPA", figsize = (12, 4))

plt.xlabel("Daily Usage Hours")
plt.ylabel("Academic Performance")
plt.title("HOW DAILY USAGE OF SOCIAL MEDIA AFFECTS ACADEMIC PERFORMANCE")

plt.show()
```

## Explanation

Most students with high GPAs have a low daily usage of social media while those with a higher daily usage of social media have much lower GPAs.


The scatter diagram shows a strong negative correlation. 

```python
data_df.plot.scatter(x = "Daily_Usage_Hours", y = "Sleep_Duration_Hours", figsize = (12, 4))

x = data_df["Daily_Usage_Hours"]
y = data_df["Sleep_Duration_Hours"]

m, b = np.polyfit(x, y, 1)

plt.plot(x, m*x + b, color='red')

plt.xlabel("Daily Usage Hours")
plt.ylabel("Sleep Duration Hours")
plt.title("Daily Usage Hours vs Sleep Duration")

plt.show()
```

## __COMMENT__

The scatter diagram shows that students that spend more time on social media, spend less time sleeping.

The scatter diagram shows a strong negative correlation.

# BOX PLOT

```python
sns.boxplot(data_df["Academic_Performance_GPA"])
plt.show()
```

## __EXPLANATION__

This box plot shows the distribution of students’ Academic Performance (GPA).

* The middle line in the box is the median GPA, which is about 3.45.
* The box shows where most students’ GPAs are found, roughly between 3.25 and 3.70.
* The whiskers show the wider range of normal GPA values, from about 2.65 to 4.00.
* The outliers show that few students have unusually low GPAs, around 2.0–2.5.
* Overall, the plot shows that most students are concentrated around 3.25–3.70.

# __PIE CHART__

```python
data_df['Primary_Platform'].value_counts().plot.pie(
    autopct = '%1.1f%%'
)

plt.gcf().set_facecolor("gray")
plt.title("PLATFORMS")
plt.ylabel('')
plt.show()
```

## __COMMENT__

This pie chart shows the different social media platforms used by students and the percentage of students using each platform.


# __BAR GRAPH__

```python
data_df.groupby('Gender')['Daily_Usage_Hours'].mean().plot.bar(figsize = (12, 6), color = "maroon")

plt.title("WHICH GENDER USES SOCIAL MEDIA MORE.")
plt.xlabel('GENDER')
plt.ylabel('DAILY_USAGE_HOURS')
plt.show()
```

## __COMMENT__

The bar graph shows that generally females spend the most time using social media than other gender groups.

```python
data_df.groupby('Academic_Level')['Daily_Usage_Hours'].mean().plot.bar(figsize = (12, 6), color = "maroon")

plt.title("HOW DIFFERENT ACADEMIC LEVELS USE SOCIAL MEDIA DAILY.")
plt.ylabel('DAILY USAGE HOURS')
plt.xlabel('ACADEMIC LEVEL')
plt.show()
```

## __COMMENT__

The bar graph shows that students in highschool spend more time on social media than other academic levels.
